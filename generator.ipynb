```python
import numpy as np
import pandas as pd

np.random.seed(42)
N = 200

# -------------------------------------------------------------
# 1. Fleet Registry (Strictly matches the 6 columns in UI)
# -------------------------------------------------------------
ship_types = ["Bulk Carrier", "Tanker", "Container Ship", "Fish Carrier"]
engine_types = ["Four-Stroke", "Two-Stroke", "Gas Turbine"]
status_types = ["Fair", "Good", "Critical"]
fuel_types = ["Heavy Fuel Oil (HFO)", "Ammonia", "LNG", "Hydrogen"]

vessels = []
for i in range(1, N + 1):
    s_type = np.random.choice(ship_types, p=[0.35, 0.35, 0.20, 0.10])
    
    if s_type == "Tanker":
        dwt = int(np.random.uniform(90000, 160000))
        draft = round(float(np.random.uniform(13.0, 17.0)), 1)
        fuel = np.random.choice(fuel_types, p=[0.35, 0.35, 0.20, 0.10])
        engine = np.random.choice(["Two-Stroke", "Four-Stroke"], p=[0.7, 0.3])
    elif s_type == "Bulk Carrier":
        dwt = int(np.random.uniform(60000, 120000))
        draft = round(float(np.random.uniform(11.0, 15.0)), 1)
        fuel = np.random.choice(fuel_types, p=[0.50, 0.20, 0.25, 0.05])
        engine = np.random.choice(["Two-Stroke", "Four-Stroke"], p=[0.6, 0.4])
    elif s_type == "Container Ship":
        dwt = int(np.random.uniform(15000, 50000))
        draft = round(float(np.random.uniform(8.0, 12.0)), 1)
        fuel = np.random.choice(fuel_types, p=[0.30, 0.15, 0.45, 0.10])
        engine = np.random.choice(engine_types, p=[0.4, 0.5, 0.1])
    else:  # Fish Carrier
        dwt = int(np.random.uniform(3000, 8000))
        draft = round(float(np.random.uniform(4.5, 7.0)), 1)
        fuel = np.random.choice(fuel_types, p=[0.10, 0.10, 0.30, 0.50])
        engine = np.random.choice(engine_types, p=[0.2, 0.3, 0.5])

    status = np.random.choice(status_types, p=[0.20, 0.70, 0.10])

    vessels.append({
        "Ship_Type": s_type,
        "Engine_Type": engine,
        "Maintenance_Status": status,
        "Fuel_Type": fuel,
        "Draft_meters": draft,
        "Max_Capacity": dwt
    })

pd.DataFrame(vessels).to_csv("vessel_registry.csv", index=False)
print("Generated clean vessel_registry.csv matching UI exactly.")
# -------------------------------------------------------------
# 2. Generate Voyage Demands (200 Voyages)
# -------------------------------------------------------------
route_types = ["Transoceanic", "Long-haul", "Coastal"]
weathers = ["Calm", "Moderate", "Rough"]

voyages = []
for j in range(1, N + 1):
    r_type = np.random.choice(route_types, p=[0.40, 0.40, 0.20])
    
    if r_type == "Transoceanic":
        dist = int(np.random.uniform(2500, 5000))
        cargo = int(np.random.uniform(30000, 80000))
    elif r_type == "Long-haul":
        dist = int(np.random.uniform(1200, 2400))
        cargo = int(np.random.uniform(15000, 40000))
    else:  # Coastal
        dist = int(np.random.uniform(300, 1000))
        cargo = int(np.random.uniform(2000, 8000))
        
    speed_est = 14.0
    min_hours = dist / speed_est
    deadline = int(round(min_hours * np.random.uniform(1.15, 1.45)))
    weather = np.random.choice(weathers, p=[0.30, 0.50, 0.20])

    voyages.append({
        "route_id": f"VOY-{j:02d}",
        "route_type": r_type,
        "distance_nm": dist,
        "cargo_weight_tons": cargo,
        "weather_condition": weather,
        "max_arrival_hours": deadline
    })

df_voyages = pd.DataFrame(voyages)
df_voyages.to_csv("voyage_demands.csv", index=False)
print(f"Generated voyage_demands.csv ({len(df_voyages)} rows)")
```
